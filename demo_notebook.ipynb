# Project Vulcan v0.2 — ROP software-on-demand demo

This notebook uses the same compiler, SafetyGate and FHIR generator as the API. It is a research demonstration, not clinical software.

```python
from vulcan.core.compiler import IntentCompiler
from vulcan.safety.gate import SafetyGate
from vulcan.interoperability.fhir import build_rop_screening_bundle
from vulcan.models.spec import ROPCase
```

```python
need = "Generate a screening workflow for Retinopathy of Prematurity for infants born at <30 weeks gestation, including FHIR scheduling and DICOM image routing, with clinician review."
spec = IntentCompiler().compile(need)
spec.model_dump()
```

```python
findings, deployable = SafetyGate().evaluate(spec)
{"deployable": deployable, "findings": [f.model_dump() for f in findings]}
```

```python
case = ROPCase(patient_id="SYN-ROP-001", gestational_age_weeks=27.4, birth_weight_g=920, exam_postmenstrual_age_weeks=34.1, sex="female")
bundle = build_rop_screening_bundle(spec, case)
bundle
```

## Safety interception example
The same deterministic gate blocks high-impact autonomous actions.

```python
dangerous = IntentCompiler().compile("For an ROP infant, autonomously adjust the infant oxygen flow rate from the screening result.")
danger_findings, danger_deployable = SafetyGate().evaluate(dangerous)
{"deployable": danger_deployable, "findings": [f.model_dump() for f in danger_findings]}
```
